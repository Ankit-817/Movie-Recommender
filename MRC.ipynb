```python
# Define the convert function to properly handle lists of genres
def convert(genre_list):
    # If the input is already a list, return it as is
    if isinstance(genre_list, list):
        return genre_list
    
    # If it's a string representation of a list, convert it to an actual list
    try:
        import ast
        return ast.literal_eval(genre_list)
    except:
        # If conversion fails, return an empty list or handle the error as needed
        return []

# Apply the convert function to the genres column
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
def convert1(obj):
    L = []
    counter = 0;
    for i in ast.literal_eval(obj):
        if counter != 3:
           L.append(i['name'])
           counter += 1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert1)
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies['overview'].apply(lambda x: x.split())
```

```python
movies['genres'] = movies['genres'].apply(lambda x: [i['name'].replace(" ","") if isinstance(i, dict) and 'name' in i else i for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i['name'].replace(" ","") if isinstance(i, dict) and 'name' in i else i for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i['name'].replace(" ","") if isinstance(i, dict) and 'name' in i else i for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i['name'].replace(" ","") if isinstance(i, dict) and 'name' in i else i for i in x])
```

```python
# Convert lists to strings before concatenation
movies['tags'] = (
    movies['overview'] + ' ' + 
    movies['genres'].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x)) + ' ' + 
    movies['keywords'].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x)) + ' ' + 
    movies['cast'].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x)) + ' ' + 
    movies['crew'].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x))
)
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
!pip install scikit-learn
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features = 5000, stop_words = 'english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
sorted_list = sorted(list(enumerate(similarity[0])), reverse=True, key=lambda x: x[1])[1:6]
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
    
```

```python
recommend('Batman')
```
